```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Grafiklerin daha net görünmesi için stil ayarları
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Seaborn kütüphanesinin içindeki hazır 'penguins' (Penguenler) veri setini yüklüyoruz
df = sns.load_dataset("penguins")

print("Veri seti başarıyla yüklendi!")
```

```python
# İlk 5 satıra bakalım
display(df.head())

# Veri seti boyutu
print(f"\nVeri seti boyutu: {df.shape[0]} satır, {df.shape[1]} sütun\n")

# Sütun bilgileri ve veri tipleri
df.info()
```

```python
# 1. Hangi sütunda kaç tane eksik değer Var?
print("--- EKSİK DEĞER SAYILARI ---")
print(df.isnull().sum())

# 2. Eksik veriye sahip satırları veri setinden çıkaralım (Doldurma yerine temizleme yapıyoruz)
df_clean = df.dropna().copy()

print("\n--- TEMİZLENDİKTEN SONRA ---")
print(f"Eski boyut: {df.shape}")
print(f"Yeni boyut: {df_clean.shape}")
print(f"Temizlenen toplam satır sayısı: {len(df) - len(df_clean)}")
```

```python
# 1. Sayısal değişkenlerin istatistiksel özeti
print("--- İSTATİSTİKSEL ÖZET (NUMERIC) ---")
display(df_clean.describe())

# 2. Grafik Paneli Oluşturma (1 satır, 2 sütun)
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Grafik A: Türlere göre Vücut Kütlesi (Boxplot)
sns.boxplot(ax=axes[0], data=df_clean, x="species", y="body_mass_g", palette="Set2")
axes[0].set_title("Türlere Göre Vücut Kütlesi Dağılımı (g)", fontsize=14, fontweight="bold")
axes[0].set_xlabel("Penguen Türü", fontsize=12)
axes[0].set_ylabel("Vücut Kütlesi (g)", fontsize=12)

# Grafik B: Gaga Uzunluğu vs Yüzgeç Uzunluğu (Scatter Plot)
sns.scatterplot(ax=axes[1], data=df_clean, x="bill_length_mm", y="flipper_length_mm", hue="species", style="sex", s=80, palette="Set2")
axes[1].set_title("Gaga Uzunluğu vs. Yüzgeç Uzunluğu", fontsize=14, fontweight="bold")
axes[1].set_xlabel("Gaga Uzunluğu (mm)", fontsize=12)
axes[1].set_ylabel("Yüzgeç Uzunluğu (mm)", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# 1. Sadece sayısal sütunları seçip korelasyon matrisini hesaplayalım
numeric_cols = df_clean.select_dtypes(include=['float64', 'int64'])
correlation_matrix = numeric_cols.corr()

# 2. Isı Haritası (Heatmap) Grafiği
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Sayısal Değişkenler Arasındaki Korelasyon Matrisi", fontsize=14, fontweight="bold")
plt.show()
```
